# Problem 1a

```python
import numpy as np
import matplotlib.pyplot as plt

Ts = 5e-6
L = 50e-6
Vo = 50.0

base_scale_Vo = (Vo * Ts) / L

D = np.linspace(0.001, 0.999, 1000) # (0 < D < 1)

delta_iL_buck_a =  (Vo * Ts) / L * (1.0 - D)
delta_iL_boost_a =  (Vo * Ts) / L * D * (1.0 - D)
delta_iL_buckboost_a =  (Vo * Ts) / L * (1.0 - D)


plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, ax1 = plt.subplots(1, 1, figsize=(7, 5.2), dpi=1500)

ax1.plot(D, delta_iL_buck_a, label=r'Buck: $\frac{V_o(1-D)T_s}{L}$', color='#1f77b4', lw=2.2)
ax1.plot(D, delta_iL_boost_a, label=r'Boost: $\frac{V_o D(1-D)T_s}{L}$', color='#ff7f0e', lw=2.2)
ax1.plot(D, delta_iL_buckboost_a, label=r'Buck-Boost: $\frac{V_o(1-D)T_s}{L}$', color='#2ca02c', lw=2.0, linestyle='--')

ax1.set_title(r'(a) $\Delta I_L$ vs. $D$ with $V_o = 50\mathrm{V},  T_s = 5\mu\mathrm{s},  L = 50\mu\mathrm{H}$', fontsize=11, pad=10)
ax1.set_xlabel('Duty Ratio ($D$)', fontsize=11)
ax1.set_ylabel(r'Inductor Current Ripple $\Delta I_L$ (A)', fontsize=11)
ax1.set_xlim(0, 1)
ax1.set_ylim(0, 5.5)
ax1.grid(True, linestyle=':', alpha=0.7)
ax1.legend(frameon=True, fontsize=10, loc='upper right')

plt.tight_layout()
plt.show()
```
